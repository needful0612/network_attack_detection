```python
import polars as pl

# Path to your massive combined file
BIG_PARQUET = "../data/combined_kitsune.parquet"
WARMUP_PATH = "../data/kitnet_warmup.parquet"

# 1. Scan the file (Lazy)
# 2. Filter for benign rows only
# 3. Take exactly 55,000 (5k for Feature Mapping + 50k for Anomaly Detection)
# 4. Stream it directly to a new smaller parquet file
(
    pl.scan_parquet(BIG_PARQUET)
    .filter(pl.col("target") == 0)
    .head(55000)
    .sink_parquet(WARMUP_PATH)
)

print(f"Done! Created warmup file at {WARMUP_PATH}")
```
